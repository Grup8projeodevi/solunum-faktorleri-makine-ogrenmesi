# Adım 3: Keşifsel Veri Analizi (EDA) ve Korelasyon
Eksik verileri doldurulmuş veri setindeki özniteliklerin birbiriyle olan ilişkisini inceleyip grafiği dışa aktarıyoruz.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("3/5: Korelasyon Matrisi çiziliyor...")

# Analize hazır veriyi okuyoruz
df_temiz = pd.read_csv('../data/analize_hazir_veri.csv')

plt.figure(figsize=(8, 6))
sns.heatmap(df_temiz.corr(), annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title("Değişkenler Arası Korelasyon Matrisi")

# Grafiği otomatik olarak analysis klasörüne kaydediyoruz
plt.savefig('../analysis/korelasyon_matrisi.png', dpi=300, bbox_inches='tight')
print("Grafik 'analysis/korelasyon_matrisi.png' olarak kaydedildi.")
plt.show()
```
