# Adım 1: Veri Yükleme ve Temizleme
Bu aşamada ham veri seti yüklenecek, analiz için hedef değişkenler seçilecek ve geçersiz (bilmiyorum/reddedildi) yanıtlar temizlenecektir.

```python
import pandas as pd
import numpy as np

print("1/5: Veri yükleniyor ve temizleniyor...")

# Doğru klasör yolundan ham veriyi okuyoruz
df = pd.read_csv('../data/brfss_survey_data_2024.csv')
df_grup8 = df[['_STATE', 'ASTHMA3', 'CHCCOPD3', 'EXERANY2']].copy()

# Geçersiz kodları boş değer (NaN) yapıyoruz
for kolon in ['ASTHMA3', 'CHCCOPD3', 'EXERANY2']:
    df_grup8[kolon] = df_grup8[kolon].replace([7, 9, 77, 99], np.nan)

print("Temizlik sonrası eksik veri durumları:")
print(df_grup8.isnull().sum())

# 2. adıma aktarmak için ara veriyi kaydediyoruz
df_grup8.to_csv('../data/adimbasi_temiz_veri.csv', index=False)
print("\nAra veri 'adimbasi_temiz_veri.csv' olarak kaydedildi.")
```
