# Adım 2: Eksik Veri Doldurma (KNN Imputation)
Bu aşamada, bir önceki adımdan gelen boşluklar KNNImputer kullanılarak doldurulacak ve hedef değişken ikili (0-1) sistemine çekilecektir.

```python
import pandas as pd
import numpy as np
from sklearn.impute import KNNImputer

print("2/5: KNN Imputation yapılıyor...")

# Bir önceki adımın çıktısını okuyoruz
df_grup8 = pd.read_csv('../data/adimbasi_temiz_veri.csv')

# Örneklem alıp KNN uyguluyoruz
df_sample = df_grup8.head(10000).copy()
imputer = KNNImputer(n_neighbors=5)
df_temiz = pd.DataFrame(imputer.fit_transform(df_sample), columns=df_grup8.columns)

# Sınıfları 1 (Astım) ve 0 (Sağlıklı) olarak standartlaştırıyoruz
df_temiz['ASTHMA3'] = df_temiz['ASTHMA3'].round().apply(lambda x: 1 if x == 1 else 0)

print("Eksik veriler dolduruldu ve sınıflar standartlaştırıldı.")

# Analize hazır ana veri setini kaydediyoruz
df_temiz.to_csv('../data/analize_hazir_veri.csv', index=False)
print("Ana veri 'analize_hazir_veri.csv' olarak kaydedildi.")
```
